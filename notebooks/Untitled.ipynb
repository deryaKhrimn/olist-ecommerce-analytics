```python
import pandas as pd
import numpy as np

# Siparişler ve Müşteriler Tabloları Yüklendi.
df_orders = pd.read_csv('../data/raw/olist_orders_dataset.csv')
df_customers = pd.read_csv('../data/raw/olist_customers_dataset.csv')

# Orders tablosunun ilk 10 satırı
display(df_orders.head(10))

# Veri tipleri ve eksik veri kontrolü
print("--- Orders Tablosu Bilgileri ---")
df_orders.info()

print("\n--- Orders Tablosu Eksik Veri Sayısı ---")
print(df_orders.isnull().sum())


```

```python
status_distribution = df_orders['order_status'].value_counts(normalize=True) * 100
print(status_distribution.round(2))
# % olarak gösterilmesi için 100 ile çarptık burada sipariş durumlarının yüzdelik durumlarına bakıldı
```
